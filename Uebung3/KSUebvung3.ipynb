```python
import sympy as sp

# Define the matrix
M = sp.Matrix([ [3.5, -1, 0],
                [-3, 3, -3],
                [-0.5, -2, 3]])

# Calculate eigenvectors for eigenvalue 0
# returns list of (eigenvalue, multiplicity, [vectors])
right_eigenvectors = M.nullspace()
left_eigenvectors = M.T.nullspace()

sp.pprint(right_eigenvectors)
sp.pprint(left_eigenvectors)


v = sp.Matrix([0.4, 1.4, 1])
v_normalized = v.normalized()
sp.pprint(v_normalized)
```
