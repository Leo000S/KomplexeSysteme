```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('default')

#generate 10.000 pseudo random numbers between 0 and 9 (a)
size = 10000
rng = np.random.default_rng()
rints = rng.integers(low=0, high=10, size=size)

#count the frequency of each number (b)
count ={}                                           
for n in rints:
    count[n] = count.get(n,0)+1


# Create the bar chart (b)
x = np.arange(10)
c = np.array([count[n] for n in range(10)])/size
plt.bar(x, c, color='skyblue')

plt.xlabel("number")
plt.ylabel("frequency")
plt.title("histogramm of random numbers")
plt.tick_params(axis="both", which="both", direction="in")
plt.grid(axis='y', linestyle='--', alpha=0.7, zorder=0)
plt.savefig('exercise1/histogram.png')

#(c) I'd expect a standard deviation of a binomial distribution o=sqrt(N*p*(1-p))=30
# ,i.e., I expect 68% of the time the count will be 1000+/-30, 95% of the time 1000+/-60.
```
