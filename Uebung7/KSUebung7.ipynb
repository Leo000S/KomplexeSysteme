```python
# 7.2 c)

import numpy as np
import matplotlib.pyplot as plt

# f(x) = cosh(x) und  f'(x) = sinh(x)
def f(x):
    return np.cosh(x)

def df(x):
    return np.sinh(x)

# 2. Hyperparameter für das Gradientenverfahren festlegen
x_0 = 1.0           # Startwert
alpha = 1         # Optimale Lernrate basierend auf der lokalen Krümmung (a=1) um das Minimum
max_iter = 10       # Anzahl der Iterationsschritte
tolerance = 1e-6    # Abbruchkriterium, falls sich x kaum noch ändert

x_k = x_0
print("\n \n")
print(f"Startwert x_0 = {x_k:.6f}, f(x_0) = {f(x_k):.6f}\n")
x_history = [x_0]
y_history = [f(x_0)]
for k in range(max_iter):
    # Der Gradientenschritt: x_{k+1} = x_k - alpha * f'(x_k)
    grad = df(x_k)
    x_next = x_k - alpha * grad

    # Werte speichern
    x_history.append(x_next)
    y_history.append(f(x_next))

    print(f"Schritt {k+1}: x_{k+1} = {x_next:.6f}, f(x_{k+1}) = {f(x_next):.6f}")

    # Überprüfung auf Konvergenz
    if abs(x_next - x_k) < tolerance:
        print(f"\nKonvergenz erreicht nach {k+1} Schritten!")
        break

    x_k = x_next


# 4. Visualisierung des Verlaufs
plt.figure(figsize=(10, 5))

# Zeichne die cosh(x) Funktion
x_vals = np.linspace(-1.5, 1.5, 400)
plt.plot(x_vals, f(x_vals), label=r'$f(x) = \cosh(x)$', color='blue', zorder=1)

# Zeichne die Iterationsschritte ein
plt.scatter(x_history, y_history, color='red', marker='o', s=50, zorder=2, label='iteration steps')
plt.plot(x_history, y_history, color='red', linestyle='--', alpha=0.6, zorder=2)

# Start- und Endpunkt hervorheben
plt.scatter(x_history[0], y_history[0], color='green', marker='X', s=150, zorder=3, label=r'start ')
plt.scatter(x_history[-1], y_history[-1], color='gold', marker='*', s=200, zorder=3, label=r'minimum')

plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```
