```python
import sympy as sp

# definition of p1, p2 and vectors
p1 = sp.Rational(1, 2)
p2 = sp.Rational(1, 2)

psi1 = sp.Matrix([1/sp.sqrt(2), 1/sp.sqrt(2)])
psi2 = sp.Matrix([1, 0])

# Density matrix
rho = sp.simplify(p1 * (psi1 * psi1.H) + p2 * (psi2 * psi2.H))
print("Rho:",sp.latex(rho))

# spectral decomposition
eigen_data = rho.eigenvects()

lambda1, _, v1_list = eigen_data[0]
lambda2, _, v2_list = eigen_data[1]

k1 = v1_list[0].normalized()
k2 = v2_list[0].normalized()
k1 = sp.simplify(k1)
k2 = sp.simplify(k2)


print("\nSpectral decomposition:",)
print(f"lambda_1 = {lambda1}")
print(f"lambda_2 = {lambda2}")
print(sp.latex(k1), sp.latex(k2))

# coefficient matrix
c = sp.Matrix.zeros(2, 2)

c[0, 0] = sp.simplify(sp.sqrt(p1) * (k1.H * psi1)[0] / sp.sqrt(lambda1))
c[0, 1] = sp.simplify(sp.sqrt(p1) * (k2.H * psi1)[0] / sp.sqrt(lambda2))
c[1, 0] = sp.simplify(sp.sqrt(p2) * (k1.H * psi2)[0] / sp.sqrt(lambda1))
c[1, 1] = sp.simplify(sp.sqrt(p2) * (k2.H * psi2)[0] / sp.sqrt(lambda2))

print("\nCoefficient matrix:",)
print(sp.latex(c))

T = sp.Matrix([[sp.Abs(c[0,0])**2, sp.Abs(c[0,1])**2],
               [sp.Abs(c[1,0])**2, sp.Abs(c[1,1])**2]])
T = sp.simplify(T)

print("\nT matrix:")
print(sp.latex(T))

# Verification
p_vec = sp.simplify(T * sp.Matrix([lambda1, lambda2]))
print("\np vector check:")
print(sp.latex(p_vec))
```
