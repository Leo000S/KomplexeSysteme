```python
import sympy as sp
lam = 0
D, lam, mu = sp.symbols('D lambda mu')

L2 = sp.Matrix([
    [0, 0, 0, -mu],
    [0, (D+lam), -D, 0],
    [0, -D, (D+lam), 0],
    [0, -lam, -lam, mu]
])

I2 = sp.eye(2)

L3_1 = sp.kronecker_product(I2, L2)

L3_2 = sp.kronecker_product(L2, I2)

L3 = L3_1 + L3_2

sp.pprint(L3)
print(sp.latex(L2), "\n")
print(sp.latex(L3), "\n")
```

```python
import sympy as sp

D, mu = sp.symbols('D mu')

L2 = sp.Matrix([
    [0, 0, 0,-mu],
    [0, D, -D, 0],
    [0, -D, D, 0],
    [0, 0, 0, mu]
])

I2 = sp.eye(2)

L3_1 = sp.kronecker_product(I2, L2)

L3_2 = sp.kronecker_product(L2, I2)

L3 = L3_1 + L3_2

sp.pprint(L3)
print(sp.latex(L3), "\n")

EV = L3.eigenvals()
EVec = L3.eigenvects()

print(EV, "\n")
print(EVec, "\n")
```
