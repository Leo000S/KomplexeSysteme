```python
#task 11.1 (a)
import sympy as sp

i = sp.I
sqrt2 = sp.sqrt(2)

# define the state vector
psi = sp.Matrix([
    sp.Rational(1, 5),
    (2 * i * sqrt2) / 5,
    sp.Rational(-4, 15),
    (-8 * i * sqrt2) / 15
])

# compute rho (conjugate transpose in SymPy is .H)
rho = psi * psi.H

# Print as latex code
print("latex code")
print(sp.latex(rho))

# purity
rho_squared = rho**2
purity = rho_squared.trace()

print("purity")
print(f"Tr(ρ²) = {purity}")
```
