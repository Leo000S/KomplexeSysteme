```python
#task 11.1 (a)
import sympy as sp

i = sp.I
sqrt2 = sp.sqrt(2)

# define the state vector
psi = sp.Matrix([
    sp.Rational(1, 5),
    (2 * i * sqrt2) / 5,
    sp.Rational(-4, 15),
    (-8 * i * sqrt2) / 15
])

# compute rho (conjugate transpose in SymPy is .H)
rho = psi * psi.H

# Print as latex code
print("latex code")
print(sp.latex(rho))

# purity
rho_squared = rho**2
purity = rho_squared.trace()

print("purity")
print(f"Tr(ρ²) = {purity}")
```

```python
#11.1 (b)
import sympy as sp

i = sp.I
sqrt2 = sp.sqrt(2)

psi = sp.Matrix([
    sp.Rational(1, 5),
    (2 * i * sqrt2) / 5,
    sp.Rational(-4, 15),
    (-8 * i * sqrt2) / 15
])
rho = psi * psi.H

rho_L = sp.Matrix([
    [rho[0,0] + rho[1,1], rho[0,2] + rho[1,3]],
    [rho[2,0] + rho[3,1], rho[2,2] + rho[3,3]]
])

rho_R = sp.Matrix([
    [rho[0,0] + rho[2,2], rho[0,1] + rho[2,3]],
    [rho[1,0] + rho[3,2], rho[1,1] + rho[3,3]]
])

print("rho_L")
print(sp.latex(rho_L))
print("\n rho_R")
print(sp.latex(rho_R))
```
